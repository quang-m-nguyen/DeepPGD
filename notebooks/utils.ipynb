```python
import pandas as pd
import os

def calculate_gc_percent(sequence):
    """
    Calculate the GC content (percentage) of a DNA sequence.
    
    Args:
    sequence (str): The DNA sequence to analyze.
    
    Returns:
    float: The GC content as a percentage.
    """
    # Convert the sequence to uppercase to handle both cases
    sequence = sequence.upper()
    
    # Count the number of G and C nucleotides
    gc_count = sequence.count('G') + sequence.count('C')
    
    # Calculate the total length of the sequence
    total_length = len(sequence)
    
    # Calculate the GC percentage
    if total_length > 0:
        gc_percent = (gc_count / total_length) * 100
    else:
        gc_percent = 0
    
    return gc_percent

# File path
input_file = '../6mA/6mA_A.thaliana/train.tsv'

# Read the TSV file
df = pd.read_csv(input_file, sep='\t')

# Calculate GC content for each sequence in the 'text' column
df['GC'] = df['text'].apply(calculate_gc_percent)

# Create the output filename
file_name, file_extension = os.path.splitext(input_file)
output_file = f"{file_name}_gc{file_extension}"

# Save the updated DataFrame to a new TSV file
df.to_csv(output_file, sep='\t', index=False)

print(f"File saved as: {output_file}")
```

```python
def calculate_average_gc(file_path):
    sum_gc_label_1 = 0
    count_label_1 = 0
    sum_gc_label_0 = 0
    count_label_0 = 0

    with open(file_path, 'r') as file:
        next(file)  # Skip the header row
        for line in file:
            parts = line.strip().split('\t')
            if len(parts) >= 4:  # Ensure the line has enough columns
                label = int(parts[1])
                gc = float(parts[3])
                
                if label == 1:
                    sum_gc_label_1 += gc
                    count_label_1 += 1
                elif label == 0:
                    sum_gc_label_0 += gc
                    count_label_0 += 1

    avg_gc_label_1 = sum_gc_label_1 / count_label_1 if count_label_1 > 0 else 0
    avg_gc_label_0 = sum_gc_label_0 / count_label_0 if count_label_0 > 0 else 0

    return avg_gc_label_1, avg_gc_label_0

# Call the function with the file path
file_path = '../4mC/4mC_F.vesca/train_gc.tsv'
avg_gc_1, avg_gc_0 = calculate_average_gc(file_path)

print(f"Average GC for label 1: {avg_gc_1:.2f}")
print(f"Average GC for label 0: {avg_gc_0:.2f}")
```
